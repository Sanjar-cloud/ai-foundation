```python
data_with_missing = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Ferrari", None, "Ford"],
    "counts": [3, 7, None, 8, 1, 9, None]
}
```

```python
df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
# 13. isnull() — Bo'sh qiymatlarni aniqlash
print(df_missing.isnull())
```

```python
# 14. isnull().sum() — Bo'sh qiymatlarni sanash
print(df_missing.isnull().sum())
```

```python
# 15. dropna() — Bo'sh qiymati bor qatorlarni o'chirish
print(df_missing.dropna())
```

```python
# 16. dropna(subset) — Faqat cars bo'sh bo'lgan qatorlarni o'chirish
print(df_missing.dropna(subset=["cars"]))
```

```python
# 17. fillna() — Bo'sh qiymatlarni to'ldirish
print(df_missing.fillna({
    "cars": "Unknown",
    "counts": 0
}))
```

```python
# 18. DataFrame'ni yangilash
df_clean = df_missing.fillna({
    "cars": "Unknown",
    "counts": 0
})
```

```python
print(df_clean)
```
